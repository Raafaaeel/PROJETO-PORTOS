```python
import pandas as pd

# Define o nome do arquivo de entrada e saída
arquivo_entrada = '/content/2025Carga.txt'
arquivo_saida = 'cargas_desembarcadas.txt'

try:
    # Lê o arquivo CSV (usando ';' como separador, conforme visto no arquivo)
    df = pd.read_csv(arquivo_entrada, sep=';')

    # Filtra o DataFrame para manter apenas as linhas
    # onde a coluna "Sentido" é igual a "Desembarcados"
    df_filtrado = df[df['Sentido'] == 'Desembarcados']

    # Salva o resultado em um novo arquivo CSV, mantendo o separador ';'
    # index=False evita que o pandas adicione uma nova coluna de índice
    df_filtrado.to_csv(arquivo_saida, sep=';', index=False)

    print(f"Arquivo filtrado com sucesso!")
    print(f"Os dados foram salvos em: {arquivo_saida}")

except FileNotFoundError:
    print(f"Erro: O arquivo '{arquivo_entrada}' não foi encontrado.")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
```

```python
import pandas as pd

# Define o nome do arquivo de entrada e o novo arquivo de saída
arquivo_entrada = '/content/cargas_desembarcadas.txt'
arquivo_saida = 'cargas_filtradas_operacao.txt'

# Lista de valores que queremos REMOVER
valores_para_excluir = [
    "Apoio",
    "Baldeação de Carga Nacional",
    "Interior",
    "Transferência Interna"
]

try:
    # Lê o arquivo original
    df = pd.read_csv(arquivo_entrada, sep=';')

    # Filtra o DataFrame, mantendo apenas as linhas onde
    # o valor da coluna "Tipo Operacao da Carga" NÃO ESTÁ (~) na lista
    df_filtrado = df[~df['Tipo Operação da Carga'].isin(valores_para_excluir)]

    # Salva o resultado em um novo arquivo
    df_filtrado.to_csv(arquivo_saida, sep=';', index=False)

    print(f"Arquivo filtrado com sucesso!")
    print(f"Registros com operações indesejadas foram removidos.")
    print(f"Os dados foram salvos em: {arquivo_saida}")

except FileNotFoundError:
    print(f"Erro: O arquivo '{arquivo_entrada}' não foi encontrado.")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
```
